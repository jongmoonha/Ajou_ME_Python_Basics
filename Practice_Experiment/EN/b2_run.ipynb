# Practice — Running Experiments b. Running a script from outside

| Item | This approach |
|------|---------------|
| Experiment code | One script, `b1_train.py` |
| Where you change values | Arguments after the run command — `--lr 0.01` |
| Editing the code | Not needed when the values change |

- Script : a `.py` file that runs from top to bottom in one go

---
# 1. `b1_train.py` — code that receives values from outside

- Open `b1_train.py` in VS Code
- Command-line argument : a value written after the run command and handed to the program — `--lr 0.01`
- `argparse` : reads the command-line arguments into `args`
- Argument names : a word starts with `--` — `--lr`. A one-letter name takes a single `-` — `-h`

| | Dictionary | `argparse` |
|---|------------|------------|
| Define a value | `'lr': 0.001` | `parser.add_argument('--lr', type=float, default=0.001)` |
| Read a value | `config['lr']` | `args.lr` |

- `type` : converts the text `0.01` typed in the command into a number
- `default` : the value used when the argument is not given

---
# 2. Running it from a notebook

| What it does | Command |
|--------------|---------|
| Run with the defaults | `%run b1_train.py` |
| Run with a given value | `%run b1_train.py --lr 0.01` |
| List the arguments | `%run b1_train.py -h` |

- Magic command : a notebook-only command that starts with `%`
- `%run` : runs a Python file inside this notebook → the output stays below the cell

````python
%run b1_train.py
````

````python
%run b1_train.py --act sigmoid --optim sgd --lr 0.01 --batch_size 128 --exp_name test
````

---
# 3. Running several times with different values

- `{...}` : inserts the value of a Python variable into the command
- Give each run a different `--exp_name` → the files in `results/` do not overwrite each other

````python
lr_list = ['0.1', '0.01', '0.001']
for lr in range(len(lr_list)):
    print(f"Learning rate: {lr_list[lr]}")
    %run b1_train.py --lr {lr_list[lr]} --exp_name test{lr}
    print("\n")
````

---
# 4. Running it from a terminal

- Terminal : a window where you type commands — VS Code menu `Terminal` → `New Terminal`
- Shell : the program in the terminal that takes your commands and runs them — PowerShell, bash

| Where it runs | Command | Output |
|---------------|---------|--------|
| Notebook cell | `%run b1_train.py --lr 0.01` | Below the cell — prints and figures |
| Notebook cell | `!python b1_train.py --lr 0.01` | Below the cell — prints only |
| Terminal | `python b1_train.py --lr 0.01` | In the terminal |

- `!` : runs a terminal command from a notebook cell
- After `%run`, the variables of `b1_train.py` (`args` and others) remain in this notebook — after a run in the terminal they do not
- Batch file `b3_run.bat` · shell script `b3_run.sh` : files that list terminal commands in order → they all run in one go

```
python b1_train.py --lr 0.1   --exp_name test0_bat
python b1_train.py --lr 0.01  --exp_name test1_bat
python b1_train.py --lr 0.001 --exp_name test2_bat
```

| Terminal | Run | Experiment names |
|----------|-----|------------------|
| Windows | `.\b3_run.bat` | `test0_bat`, `test1_bat`, `test2_bat` |
| macOS · Linux | `bash b3_run.sh` | `test0_sh`, `test1_sh`, `test2_sh` |

- The terminal's current folder must be the folder that holds these files → move there first with `cd` (the command that changes folders)
