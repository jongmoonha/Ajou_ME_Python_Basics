# Practice — Running Experiments a. Changing values inside a notebook

- Hyperparameter : a value you choose before training — learning rate, batch size, and so on. These are the values changed in this practice
- No model is trained — we only watch the values change and get passed along

---
# 1. Starting point — values written directly in the code (hard-coding)

```python
act = 'relu'
optim = 'adam'
lr = 0.001
batch_size = 32
exp_name = 'name'

print("Training with the following parameters:")
print(f"Activation: {act}")
print(f"Optimizer: {optim}")
print(f"Learning Rate: {lr}")
print(f"Batch Size: {batch_size}")
print(f"Experiment Name: {exp_name}")
```

| What we want | Right now |
|--------------|-----------|
| Run again with a different value | Edit the value in the cell above and run it again |
| Compare with the previous run | Running again erases the previous output |
| Try several values | Repeat edit-and-run that many times |

---
# 2. Collect the values in a dictionary and make the experiment a function

- `config` : a dictionary that holds all the values of an experiment in one place
- `train(config)` : a function that runs the experiment with the values it receives → call it again with different values

```python
config = {
    'act': 'relu',
    'optim': 'adam',
    'lr': 0.001,
    'batch_size': 32,
    'exp_name': 'name',
}
```

```python
def train(config):
    print("Training with the following parameters:")
    print(f"Activation: {config['act']}")
    print(f"Optimizer: {config['optim']}")
    print(f"Learning Rate: {config['lr']}")
    print(f"Batch Size: {config['batch_size']}")
    print(f"Experiment Name: {config['exp_name']}")

train(config)
```

---
# 3. Run several times, changing only the values (sweep)

- Inside the `for` loop only the values in `config` change → `train` is not edited
- Each run is saved as `results/<experiment name>.csv` → a record of which values were used

```python
import os
import pandas as pd

os.makedirs('results', exist_ok=True)

lr_list = [0.1, 0.01, 0.001]
for lr in range(len(lr_list)):
    print(f"Learning rate: {lr_list[lr]}")
    config['lr'] = lr_list[lr]
    config['exp_name'] = f'test{lr}'
    train(config)
    print("\n")

    filename_df = './results/{}.csv'.format(config['exp_name'])
    config_df = pd.DataFrame(config, index=[0])
    config_df.to_csv(filename_df)
```

---
# 4. What a dictionary cannot do

- To change a value you must open this notebook and edit a cell — the values and the experiment code live in the same file
- To leave the experiment code untouched and pass in only the values **from outside** (another notebook, a terminal) → `argparse`

| Where the values live | Where you change them | File |
|-----------------------|-----------------------|------|
| Dictionary | Inside the code — edit a cell | `a_notebook.ipynb` |
| `argparse` | Outside the code — pass them as arguments when running | `b1_train.py`, `d1_setup.ipynb` |
| Config file (YAML) | Outside the code — write them in a file and pass the file name when running | `c1_train_yaml.py`, `configs_yaml/` |
