# Practice — Running Experiments (d1) The Base Experiment Notebook

- The same code as `b1_train.py`, kept as a notebook → you write the experiment code while running it cell by cell
- Run this notebook as it is → a run with the defaults
- For other values → `d2_run.ipynb` runs this notebook

| | `b1_train.py` | This notebook |
|---|---------------|---------------|
| Reading the arguments | `parser.parse_args()` | `parser.parse_known_args()` |

- `sys.argv` : the list that holds what was written in the run command → `argparse` reads the arguments from here
- In a notebook, `sys.argv` holds the arguments given to the kernel (the program that runs the cells) → `parse_args()` stops with an error when it finds an argument it does not know
- `parse_known_args()` : skips the arguments it does not know

---
# 1. Defining the Arguments and Checking the Values

```python
import argparse
import os
import pandas as pd

parser = argparse.ArgumentParser()

# ① Define every argument and set its default with default=
parser.add_argument('--act',        type=str,   default='relu',        help='activation function')
parser.add_argument('--optim',      type=str,   default='adam',        help='optimizer')
parser.add_argument('--lr',         type=float, default=0.001,         help='learning rate')
parser.add_argument('--batch_size', type=int,   default=32,            help='batch size')
parser.add_argument('--exp_name',   type=str,   default='name',        help='exp name')

# ② Read the arguments in sys.argv - unknown ones are skipped
args, unknown = parser.parse_known_args()

# ③ Check the values
print("Training with the following parameters:")
print(f"Activation: {args.act}")
print(f"Optimizer: {args.optim}")
print(f"Learning Rate: {args.lr}")
print(f"Batch Size: {args.batch_size}")
print(f"Experiment Name: {args.exp_name}")
```

---
# 2. Saving the Result

- Saves the values of this run to `results/<experiment name>.csv`

```python
# ④ Save the result
os.makedirs('results', exist_ok=True)
filename_df = './results/{}.csv'.format(args.exp_name)
args_df = pd.DataFrame(vars(args), index=[0])
args_df.to_csv(filename_df)
```
