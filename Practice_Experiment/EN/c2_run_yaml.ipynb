# Practice — Running Experiments (c) Running with Config Files

| Item | This approach |
|------|---------------|
| Experiment code | One script, `c1_train_yaml.py` |
| Where you change values | Config files — the YAML files in `configs_yaml/` |
| Editing the code | Not needed when the values change |

- Config file: a file that lists the values for an experiment
- YAML: a format for config files — one `name: value` per line (put a space after the `:`)

---
# 1. `c1_train_yaml.py` — Code That Reads a Config File

- Open `c1_train_yaml.py` in VS Code
- Same code as `b1_train.py`, with three additions

| Where | Code | What it does |
|-------|------|--------------|
| At the top | `import yaml` | Imports the module that reads YAML files |
| Argument definitions | `parser.add_argument('--config', ...)` | Takes the config file name |
| After reading the arguments | `if args.config is not None:` and the three lines below it | Sets the values in the file as the `default` values, then reads the arguments again |

- `yaml.safe_load(f)` : reads the `name: value` lines of the file into a dictionary
- If an error says there is no module named `yaml` → run `pip install pyyaml` in a terminal

---
# 3. Running Several Times with Different Config Files

- One config file per run → the files remain as a record of the runs
- A `for` loop runs the script with each config file in turn

````python
config_list = ['configs_yaml/test0.yaml', 'configs_yaml/test1.yaml', 'configs_yaml/test2.yaml']
for config in config_list:
    print(f"Config: {config}")
    %run c1_train_yaml.py --config {config}
    print("\n")
````

---
# 4. Running It from a Terminal

- The command is the same in a terminal — `python c1_train_yaml.py --config configs_yaml/test.yaml`
- `c3_run_yaml.bat`, `c3_run_yaml.sh` : files listing the commands that run the script with each of the three config files

```
python c1_train_yaml.py --config configs_yaml/test0.yaml --exp_name test0_yaml_bat
python c1_train_yaml.py --config configs_yaml/test1.yaml --exp_name test1_yaml_bat
python c1_train_yaml.py --config configs_yaml/test2.yaml --exp_name test2_yaml_bat
```

| Terminal | Run | Experiment names |
|----------|-----|------------------|
| Windows | `.\c3_run_yaml.bat` | `test0_yaml_bat`, `test1_yaml_bat`, `test2_yaml_bat` |
| macOS / Linux | `bash c3_run_yaml.sh` | `test0_yaml_sh`, `test1_yaml_sh`, `test2_yaml_sh` |

- An argument added to the command takes priority over the config file → the `--exp_name` in the commands above overrides the experiment name written in the config file

---
# 5. Keep in Mind

- A wrong name in the config file (`learning_rate: 0.5`) raises no error and the run uses the `default` → check the printed values
