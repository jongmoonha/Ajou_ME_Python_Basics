# Practice — Running Experiments d2. Running a notebook from outside

| Item | This approach |
|------|---------------|
| Experiment code | One notebook, `d1_setup.ipynb` |
| Where you change values | This notebook — put the arguments into `sys.argv` |
| Output | Piles up below the cells of this notebook, in the order of the runs |

---
# 1. `sys.argv` — where the command-line arguments arrive

| Terminal command | `sys.argv` as filled by Python |
|------------------|--------------------------------|
| `python b1_train.py` | `['b1_train.py']` |
| `python b1_train.py --lr 0.01` | `['b1_train.py', '--lr', '0.01']` |

- `argparse` reads `sys.argv` from the second item on — the first item is the place for the file name
- Arguments written after `%run` are not passed to a notebook → fill `sys.argv` yourself, then `%run`
- Any value will do for the first item — here `'ignored'`

```python
import sys
sys.argv = ['ignored']      # no arguments → all defaults
%run ./d1_setup.ipynb
```

---
# 2. Running with given values

- Write each argument name and each value as a separate string item

```python
sys.argv = [
    'ignored',
    '--act', 'sigmoid',
    '--optim',  'sgd',
    '--lr',    '0.01',
    '--batch_size','128',
    '--exp_name', 'test',
]

%run ./d1_setup.ipynb
```

---
# 3. Running several times with different values

- Inside the `for` loop only `sys.argv` changes → `d1_setup.ipynb` is not edited

```python
lr_list = ['0.1', '0.01', '0.001']
for lr in range(len(lr_list)):
    print(f"Learning rate: {lr_list[lr]}")

    sys.argv = [
    'ignored',
    '--lr',    lr_list[lr],
    '--exp_name', f'test{lr}',
    ]
    %run ./d1_setup.ipynb
    print("\n")
```

---
# 4. How this differs from running a script

| | `%run b1_train.py --lr 0.01` | `sys.argv` + `%run ./d1_setup.ipynb` |
|---|------------------------------|---------------------------------------|
| Experiment code | A `.py` file | A notebook — built by running it cell by cell |
| A misspelled argument name | Raises an error and that run does not happen | Is skipped and the run uses the `default` |
| Running with `python` in a terminal | Works | Does not work |

- After each run, check that the printed values match the values you passed
