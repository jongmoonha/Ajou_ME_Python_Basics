# Practice — 실험 실행 d2. 노트북을 밖에서 실행하기

| 항목 | 이 방식 |
|------|---------|
| 실험 코드 | `d1_setup.ipynb` 노트북 하나 |
| 값을 바꾸는 곳 | 이 노트북 — `sys.argv` 에 인자를 넣는다 |
| 출력 | 이 노트북의 셀 아래에 실험 순서대로 쌓인다 |

---
# 1. `sys.argv` — 명령줄 인자가 들어오는 곳

| 터미널 명령 | 파이썬이 채우는 `sys.argv` |
|-------------|---------------------------|
| `python b1_train.py` | `['b1_train.py']` |
| `python b1_train.py --lr 0.01` | `['b1_train.py', '--lr', '0.01']` |

- `argparse` 는 `sys.argv` 의 둘째 칸부터 읽는다 — 첫 칸은 파일 이름 자리
- `%run` 뒤에 적은 인자는 노트북에는 전달되지 않는다 → `sys.argv` 를 직접 채운 뒤 `%run`
- 첫 칸에는 아무 값이나 둔다 — 여기서는 `'ignored'`

```python
import sys
sys.argv = ['ignored']      # 인자 없음 → 모두 default
%run ./d1_setup.ipynb
```

---
# 2. 값을 지정해 실행하기

- 인자 이름과 값을 문자열로 한 칸씩 적는다

```python
sys.argv = [
    'ignored',
    '--act', 'sigmoid',
    '--optim',  'sgd',
    '--lr',    '0.01',
    '--batch_size','128',
    '--exp_name', 'test',
]

%run ./d1_setup.ipynb
```

---
# 3. 값을 바꿔 여러 번 실행하기

- `for` 문 안에서 `sys.argv` 만 바꾼다 → `d1_setup.ipynb` 는 고치지 않는다

```python
lr_list = ['0.1', '0.01', '0.001']
for lr in range(len(lr_list)):
    print(f"Learning rate: {lr_list[lr]}")

    sys.argv = [
    'ignored',
    '--lr',    lr_list[lr],
    '--exp_name', f'test{lr}',
    ]
    %run ./d1_setup.ipynb
    print("\n")
```

---
# 4. 스크립트를 실행할 때와 다른 점

| | `%run b1_train.py --lr 0.01` | `sys.argv` + `%run ./d1_setup.ipynb` |
|---|-----------------------------|---------------------------------------|
| 실험 코드 | `.py` 파일 | 노트북 — 셀 단위로 실행하며 만든다 |
| 인자 이름을 잘못 적으면 | 오류를 내고 그 실험은 실행되지 않는다 | 건너뛰고 `default` 로 실행한다 |
| 터미널에서 `python` 으로 실행 | 된다 | 안 된다 |

- 실행한 뒤 출력된 값이 넣은 값과 같은지 확인한다
- 설정 파일도 쓰려면 → `c1_train_yaml.py` 에 더해진 세 곳을 `d1_setup.ipynb` 에 옮기고, `sys.argv` 에 `'--config', '파일 이름'` 을 넣는다
