# Practice — 실험 실행 c. 설정 파일로 실행하기

| 항목 | 이 방식 |
|------|---------|
| 실험 코드 | 스크립트 `c1_train_yaml.py` 하나 |
| 값을 바꾸는 곳 | 설정 파일 — `configs_yaml/` 의 yaml 파일 |
| 코드 수정 | 값을 바꿔도 고치지 않는다 |

- 설정 파일(config file) : 실험에 쓰는 값을 적어 둔 파일
- YAML : 설정 파일을 쓰는 형식 — `이름: 값` 을 한 줄에 하나씩 적는다 (`:` 뒤에 한 칸 띄운다)

---
# 1. `c1_train_yaml.py` — 설정 파일을 읽는 코드

- VS Code 에서 `c1_train_yaml.py` 를 열어 본다
- `b1_train.py` 에 세 곳이 더해졌다

| 더해진 곳 | 코드 | 하는 일 |
|-----------|------|---------|
| 맨 위 | `import yaml` | yaml 파일을 읽는 모듈을 불러온다 |
| 인자 정의 | `parser.add_argument('--config', ...)` | 설정 파일 이름을 받는다 |
| 인자를 읽은 뒤 | `if args.config is not None:` 과 그 아래 세 줄 | 파일에 적은 값을 `default` 로 넣고, 인자를 다시 읽는다 |

- `yaml.safe_load(f)` : 파일에 적은 `이름: 값` 들을 딕셔너리로 읽는다
- `yaml` 모듈이 없다는 오류가 나오면 → 터미널에서 `pip install pyyaml`

---
# 2. 설정 파일 하나로 실행하기

- 명령 뒤에 길게 적던 값들을 파일에 옮겨 적는다 — `--act sigmoid --optim sgd --lr 0.01 --batch_size 128`
- `--config` 뒤에 파일 이름을 준다 → 파일에 적은 값이 `default` 대신 쓰인다

`configs_yaml/test.yaml`

```yaml
act: sigmoid
optim: sgd
lr: 0.01
batch_size: 128
exp_name: test_yaml
```

````python
%run c1_train_yaml.py --config configs_yaml/test.yaml
````

---
# 3. 설정 파일을 바꿔 여러 번 실행하기

- 실험 하나에 설정 파일 하나 → 파일이 실험 기록으로 남는다
- `for` 문으로 설정 파일을 차례로 실행한다

````python
config_list = ['configs_yaml/test0.yaml', 'configs_yaml/test1.yaml', 'configs_yaml/test2.yaml']
for config in config_list:
    print(f"Config: {config}")
    %run c1_train_yaml.py --config {config}
    print("\n")
````

---
# 4. 터미널에서 실행하기

- 터미널에서도 같은 명령이다 — `python c1_train_yaml.py --config configs_yaml/test.yaml`
- `c3_run_yaml.bat`, `c3_run_yaml.sh` : 설정 파일 3개를 차례로 실행하는 명령을 적어 둔 파일

```
python c1_train_yaml.py --config configs_yaml/test0.yaml --exp_name test0_yaml_bat
python c1_train_yaml.py --config configs_yaml/test1.yaml --exp_name test1_yaml_bat
python c1_train_yaml.py --config configs_yaml/test2.yaml --exp_name test2_yaml_bat
```

| 터미널 | 실행 | 실험 이름 |
|--------|------|-----------|
| Windows | `.\c3_run_yaml.bat` | `test0_yaml_bat`, `test1_yaml_bat`, `test2_yaml_bat` |
| macOS · Linux | `bash c3_run_yaml.sh` | `test0_yaml_sh`, `test1_yaml_sh`, `test2_yaml_sh` |

- 명령 뒤에 인자를 덧붙이면 그 값이 설정 파일보다 우선한다 → 위 명령의 `--exp_name` 이 설정 파일에 적은 실험 이름 대신 쓰인다
