# Practice — 실험 실행 b. 스크립트를 밖에서 실행하기

| 항목 | 이 방식 |
|------|---------|
| 실험 코드 | 스크립트 `b1_train.py` 하나 |
| 값을 바꾸는 곳 | 실행 명령 뒤의 인자 — `--lr 0.01` |
| 코드 수정 | 값을 바꿔도 고치지 않는다 |

- 스크립트(script) : 위에서 아래로 한 번에 실행하는 `.py` 파일

---
# 1. `b1_train.py` — 값을 밖에서 받는 코드

- VS Code 에서 `b1_train.py` 를 열어 본다
- 명령줄 인자(command-line argument) : 실행 명령 뒤에 적어 프로그램에 넘기는 값 — `--lr 0.01`
- `argparse` : 명령줄 인자를 읽어 `args` 에 담는다
- 인자 이름 : 낱말은 `--` 로 시작한다 — `--lr`. 한 글자 이름은 `-` 하나 — `-h`

| | 딕셔너리 | `argparse` |
|---|----------|------------|
| 값 정의 | `'lr': 0.001` | `parser.add_argument('--lr', type=float, default=0.001)` |
| 값 읽기 | `config['lr']` | `args.lr` |

- `type` : 명령에 적은 글자 `0.01` 을 숫자로 바꾼다
- `default` : 인자를 주지 않았을 때 쓰는 값

---
# 2. 노트북에서 실행하기

| 하는 일 | 명령 |
|---------|------|
| 기본값으로 실행 | `%run b1_train.py` |
| 값을 지정해 실행 | `%run b1_train.py --lr 0.01` |
| 인자 목록 보기 | `%run b1_train.py -h` |

- 매직 명령(magic command) : `%` 로 시작하는 노트북 전용 명령
- `%run` : 파이썬 파일을 이 노트북에서 실행한다 → 출력이 셀 아래에 남는다

````python
%run b1_train.py
````

````python
%run b1_train.py --act sigmoid --optim sgd --lr 0.01 --batch_size 128 --exp_name test
````

---
# 3. 값을 바꿔 여러 번 실행하기

- `{...}` : 파이썬 변수의 값을 명령에 끼워 넣는다
- 실험마다 `--exp_name` 을 다르게 준다 → `results/` 의 파일이 서로 덮어쓰지 않는다

````python
lr_list = ['0.1', '0.01', '0.001']
for lr in range(len(lr_list)):
    print(f"Learning rate: {lr_list[lr]}")
    %run b1_train.py --lr {lr_list[lr]} --exp_name test{lr}
    print("\n")
````

---
# 4. 터미널에서 실행하기

- 터미널(terminal) : 명령을 입력하는 창 — VS Code 메뉴 `Terminal` → `New Terminal`
- 셸(shell) : 터미널에서 명령을 받아 실행하는 프로그램 — PowerShell, bash

| 실행하는 곳 | 명령 | 출력 |
|-------------|------|------|
| 노트북 셀 | `%run b1_train.py --lr 0.01` | 셀 아래 — print, 그림 |
| 노트북 셀 | `!python b1_train.py --lr 0.01` | 셀 아래 — print 만 |
| 터미널 | `python b1_train.py --lr 0.01` | 터미널 화면 |

- `!` : 노트북 셀에서 터미널 명령을 실행한다
- `%run` 으로 실행하면 `b1_train.py` 의 변수(`args` 등)가 이 노트북에 남는다 — 터미널에서 실행하면 남지 않는다
- 배치 파일(batch file) `b3_run.bat` · 셸 스크립트(shell script) `b3_run.sh` : 터미널 명령을 순서대로 적어 둔 파일 → 한 번에 실행된다

```
python b1_train.py --lr 0.1   --exp_name test0_bat
python b1_train.py --lr 0.01  --exp_name test1_bat
python b1_train.py --lr 0.001 --exp_name test2_bat
```

| 터미널 | 실행 | 실험 이름 |
|--------|------|-----------|
| Windows | `.\b3_run.bat` | `test0_bat`, `test1_bat`, `test2_bat` |
| macOS · Linux | `bash b3_run.sh` | `test0_sh`, `test1_sh`, `test2_sh` |

- 터미널의 현재 폴더가 이 파일들이 있는 폴더여야 한다 → `cd` (폴더를 옮기는 명령)로 먼저 이동한다
