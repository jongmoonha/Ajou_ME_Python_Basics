# Practice — 실험 실행 a. 노트북 안에서 값 바꾸기

- 하이퍼파라미터(hyperparameter) : 학습 전에 사람이 정해 주는 값 — 학습률, 배치 크기 등. 이 실습에서 바꿔 보는 값이다
- 모델 학습은 하지 않는다 — 값이 바뀌어 전달되는 것만 본다

---
# 1. 출발점 — 값을 코드에 직접 쓴다 (하드코딩, hard-coding)

```python
act = 'relu'
optim = 'adam'
lr = 0.001
batch_size = 32
exp_name = 'name'

print("Training with the following parameters:")
print(f"Activation: {act}")
print(f"Optimizer: {optim}")
print(f"Learning Rate: {lr}")
print(f"Batch Size: {batch_size}")
print(f"Experiment Name: {exp_name}")
```

| 하고 싶은 것 | 지금은 |
|--------------|--------|
| 값을 바꿔 다시 실험한다 | 위 셀의 값을 고치고 다시 실행한다 |
| 앞 실험과 비교한다 | 다시 실행하면 앞 출력이 지워진다 |
| 여러 값으로 실험한다 | 고치고 실행하기를 그만큼 반복한다 |

---
# 2. 값을 딕셔너리에 모으고, 실험을 함수로 만든다

- `config` : 실험에 쓰는 값을 한 곳에 모은 딕셔너리
- `train(config)` : 값을 받아 실험하는 함수 → 값만 바꿔 다시 부른다

```python
config = {
    'act': 'relu',
    'optim': 'adam',
    'lr': 0.001,
    'batch_size': 32,
    'exp_name': 'name',
}
```

```python
def train(config):
    print("Training with the following parameters:")
    print(f"Activation: {config['act']}")
    print(f"Optimizer: {config['optim']}")
    print(f"Learning Rate: {config['lr']}")
    print(f"Batch Size: {config['batch_size']}")
    print(f"Experiment Name: {config['exp_name']}")

train(config)
```

---
# 3. 값만 바꿔 여러 번 실험한다 (스윕, sweep)

- `for` 문 안에서 `config` 의 값만 바꾼다 → `train` 은 고치지 않는다
- 실험마다 `results/실험이름.csv` 로 저장한다 → 어떤 값으로 실험했는지 남는다

```python
import os
import pandas as pd

os.makedirs('results', exist_ok=True)

lr_list = [0.1, 0.01, 0.001]
for lr in range(len(lr_list)):
    print(f"Learning rate: {lr_list[lr]}")
    config['lr'] = lr_list[lr]
    config['exp_name'] = f'test{lr}'
    train(config)
    print("\n")

    filename_df = './results/{}.csv'.format(config['exp_name'])
    config_df = pd.DataFrame(config, index=[0])
    config_df.to_csv(filename_df)
```

---
# 4. 딕셔너리로 할 수 없는 것

- 값을 바꾸려면 이 노트북을 열어 셀을 고쳐야 한다 — 값과 실험 코드가 한 파일에 있다
- 실험 코드는 그대로 두고 **밖에서**(다른 노트북, 터미널) 값만 넣으려면 → `argparse`

| 값을 담는 곳 | 값을 바꾸는 곳 | 파일 |
|--------------|----------------|------|
| 딕셔너리 | 코드 안 — 셀을 고친다 | `a_notebook.ipynb` |
| `argparse` | 코드 밖 — 실행할 때 인자로 준다 | `b1_train.py`, `d1_setup.ipynb` |
| 설정 파일 (YAML) | 코드 밖 — 파일에 적어 두고, 실행할 때 파일 이름을 준다 | `c1_train_yaml.py`, `configs_yaml/` |
